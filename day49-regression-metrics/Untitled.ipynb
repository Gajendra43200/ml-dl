```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('placement.csv')
```

```python
df.head()
df.shape
```

```python
plt.sactter(df['cgpa'],df)
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('Package(in lpa)')
```

```python
X = df.iloc[:,0:1]
y = df.iloc[:,-1]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(X_train,lr.predict(X_train),color='red')
plt.xlabel('CGPA')
plt.ylabel('Package(in lpa)')
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
y_pred = lr.predict(X_test)
```

```python
y_test.values
```

```python
print("MAE",mean_absolute_error(y_test,y_pred))
```

```python
print("MSE",mean_squared_error(y_test,y_pred))
```

```python
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred)))
```

```python
print("MSE",r2_score(y_test,y_pred))
r2 = r2_score(y_test,y_pred)
```

```python
# Adjusted R2 score
X_test.shape
```

```python
1 - ((1-r2)*(40-1)/(40-1-1))# Adjusted R2 score this is use when we have more than one feature in our dataset and we are add a new feature in our dataset then we can check the adjusted R2 score if it is increase then we can say that the new feature is good for our model otherwise we can say that the new feature is not good for our model. 
#R2 score is used to check the accuracy of our model and it is between 0 and 1. The closer the R2 score to 1, the better the model is. The closer the R2 score to 0, the worse the model is. The R2 score can be negative if the model is worse than the mean model. The mean model is a model that always predicts the mean of the target variable.
```
