```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
# Linear Regression
reg = LinearRegression()
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# Ridge 
reg = Ridge(alpha=0.1)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# Lasso
reg = Lasso(alpha=0.01)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# ElasticNet
reg = ElasticNet(alpha=0.005,l1_ratio=0.9)#l1_ratio is a parameter in the ElasticNet regression model that controls the balance between L1 and L2 regularization. It takes a value between 0 and 1, where a value of 0 corresponds to pure L2 regularization (Ridge regression), a value of 1 corresponds to pure L1 regularization (Lasso regression), and values in between represent a combination of both types of regularization. In this case, l1_ratio=0.9 means that the model will primarily use L1 regularization while still incorporating some L2 regularization, which can help in feature selection while also maintaining some stability in the coefficients. and alpha is the regularization strength parameter that controls the amount of shrinkage applied to the coefficients in the ElasticNet regression model. A higher alpha value will result in more regularization, which can lead to smaller coefficients and potentially better generalization on unseen data, while a lower alpha value will result in less regularization, allowing the model to fit the training data more closely but potentially leading to overfitting. In this case, alpha=0.005 indicates a relatively low level of regularization, allowing the model to fit the data more closely while still applying some shrinkage to the coefficients.
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```
