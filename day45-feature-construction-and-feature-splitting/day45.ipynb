```python
import numpy as np
import pandas as pd

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression

import seaborn as sns
```

```python
df = pd.read_csv('train.csv')[['Age','Pclass','SibSp','Parch','Survived']]
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df.head()
```

```python
X = df.iloc[:,0:4]
y = df.iloc[:,-1]
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))
#cross validation score is a function that takes a model, the data, the target, the scoring method and the number of folds for cross validation. It returns an array of scores for each fold. We take the mean of these scores to get an overall accuracy of the model.
# this is a very basic model and we can improve it by adding more features, doing feature engineering, or using a different model.
```

## Applying Feature Construction

```python
X['Family_size'] = X['SibSp'] + X['Parch'] + 1 
# we can create a new feature called family size by adding the number of siblings/spouses and the number of parents/children and adding 1 for the person themselves. This feature might be useful because it can capture the effect of having a family on survival. For example, people with a larger family size might have a higher chance of survival because they can help each other, or they might have a lower chance of survival because they have to share resources. We can test this by adding this feature to our model and seeing if it improves the accuracy.
```

```python
X.head()
```

```python
def myfunc(num):
    if num == 1:
        #alone
        return 0
    elif num >1 and num <=4:
        # small family
        return 1
    else:
        # large family
        return 2
```

```python
myfunc(4)
```

```python
X['Family_type'] = X['Family_size'].apply(myfunc)
#apply is a function that applies a function to each element of a series. In this case, we are applying the myfunc function to each element of the Family_size column. This will create a new column called Family_type that will have the values 0, 1, or 2 depending on the family size. We can then use this new feature in our model and see if it improves the accuracy.
```

```python
X.head()
```

```python
X.drop(columns=['SibSp','Parch','Family_size'],inplace=True)
# X[usecols=['Age','Pclass','Family_type']].head()
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))
```

## Feature Splitting

```python
df = pd.read_csv('train.csv')
```

```python
df.head()
```

```python
df['Name']
```

```python
df['Name'].str.split(', ', expand=True)[1].str.split('.', expand=True)[0]
#expand is a parameter that tells the split function to return a dataframe instead of a series. This way we can access the different parts of the name easily.
```

```python
df['Title'] = df['Name'].str.split(', ', expand=True)[1].str.split('.', expand=True)[0]
# split function is a string method that splits a string into a list of substrings based on a delimiter. In this case, we are splitting the Name column first by ', ' to separate the last name from the title and first name, and then we are splitting the second part by '.' to separate the title from the first name. This will create a new column called Title that will have the titles of the passengers, such as Mr, Mrs, Miss, etc. We can then use this new feature in our model and see if it improves the accuracy.
```

```python
df[['Title','Name']]
```

```python
(df.groupby('Title').mean()['Survived']).sort_values(ascending=False)
#"Group passengers by title (Mr, Mrs, Miss), compute average values, then show average survival rate for each title."
#groupby is a function that groups the data by a certain column and then we can apply an aggregation function to get the mean survival rate for each title. This can help us understand which titles had a higher chance of survival and which had a lower chance. For example, we might find that people with the title 'Mrs' had a higher survival rate than those with the title 'Mr', which could be due to various social factors at the time of the Titanic disaster.
```
